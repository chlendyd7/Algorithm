# ndarray 인덱싱과 슬라이싱

```python
import numpy as np
```

## 인덱싱
 - 파이썬 리스트와 동일한 개념으로 사용

### 1차원 배열 (1d-array, 벡터) 인덱싱

```python
x = np.arange(10)
print(x)
```

```python
#x에서 0을 출력하시오
x[0]
```

```python
#x에서 5를 출력하시오
x[5]
```

```python
#x에서 5를 100으로 바꾸어 x를 출력하시오
x[5]=100
x
```

### 2차원 배열 (2d-array) 인덱싱

```python
# 0에서 9까지 값을 갖는 1차배열 x와 2차 배열 y(2행 5열)를 만드시오.
x = np.arange(10)
y = x.reshape(2, 5)
print(x)
print(y)
```

```python
y[0]
```

```python
y[1]
```

```python
y[0,0] 
```

```python
y[1,-1] 
```

#### (list와 비교)

```python
### list에서는 3을 출력하기 위해 어떻게 하였는가?
c=[[1, 2], [3, 4], [5, 6]] 
c[1][0]
```

```python
#그럼 2d-array도 [][]으로 indexing 가능할까?
# 해보세요!

y[1][-1] 
```

### 3차원 배열 (3d-array, 텐서) 인덱싱

```python
# 3행 2열 4개
x = np.arange(24).reshape(4, 3, 2)
print(x)
```

```python
# 두번째 행렬을 indexing
x[1]
```

```python
# 두번째 행렬, 첫행을 indexing
x[1,0]
```

```python
# 두번째 행렬, 첫행, 마지막 열을 indexing
x[1,0,1]

# -1 indexing도 가능한지 확인해 볼것!
```

### 슬라이싱
 - 리스트, 문자열 slicing과 유사하게 사용
 - ,를 사용하여 각 차원별로 슬라이싱 가능

### 1차원 배열 슬라이싱

```python
x = np.arange(10)
print(x)
```

```python
x[1:7]
```

```python
#index 1부터 끝까지 슬라이싱
x[1:]
```

```python
#처음부터 끝까지 슬라이싱
x[:]
```

### 2차원 배열 슬라이싱

```python
x = np.arange(10).reshape(2, 5)
print(x)
```

```python
x[:, 1:4]    
```

```python
x[0, :2]
```

```python
# (행) 두번째 행
# (열) 두번째 열에서 4번째 열까지

x[1,1:4]
```

### 3차원 배열 슬라이싱

```python
#3행 9열 X 2개

x = np.arange(54).reshape(2, 3, 9)
print(x)
```

```python
x[:1, :2, 1:5]
```
