### (5) reshape 함수 활용
 - ndarray의 형태, 차원을 바꾸기 위해 사용
 - 원본의 형태는 바뀌지 않고 그대로 유지됨
 - 행방향 reshape: order='C' (기본값, 생략가능) 
 - 열방향 reshape: order='F' 

```python
# 열방향으로 reshape하려면?
x.reshape(5,3, <blank>)
```

```python
# x의 형태와 차원을 확인하자!
print(x)
print(x.ndim)
```

```python
# x의 차원을 실제 변환하려면 어떻게 해야할까?
<blank>=x.reshape(5, 3)
```

```python
# x의 형태와 차원을 다시 확인
print(x)
print(x.<blank>)
```
