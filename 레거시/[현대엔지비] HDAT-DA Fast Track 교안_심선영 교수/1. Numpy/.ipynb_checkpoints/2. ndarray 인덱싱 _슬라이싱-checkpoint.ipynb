### 슬라이싱
 - 리스트, 문자열 slicing과 유사하게 사용
 - ,를 사용하여 각 차원별로 슬라이싱 가능

### 1차원 배열 슬라이싱

```python
x = np.arange(10)
print(x)
```

```python
#index 1부터 끝까지 슬라이싱
x[1:<blank>]
```

```python
#처음부터 끝까지 슬라이싱
x[<blank>]
```

### 2차원 배열 슬라이싱

```python
x = np.arange(10).reshape(2, 5)
print(x)
```

```python
# (행) 두번째 행
# (열) 두번째 열에서 4번째 열까지

x[<blank>, <blank>]
```

### 3차원 배열 슬라이싱

```python
#3행 9열 X 2개

x = np.arange(54).reshape(2, 3, 9)
print(x)
```
