# Boolean Indexing

- nd-array에 조건문을 인덱스처럼 전달하여 True인 경우만 필터링

## 짝수인 경우만 필터링

```python
import numpy as np
```

- [1,100) 범위에서 10개의 정수 랜덤 추출

```python

x = np.random.randint(1, 100, size=10)
print(x)
```

- random seed 를 300으로 고정해 보자!

- 조건문 생성 :x는 짝수?
    - 조건문의 결과 = (True/False) 

```python
# 조건문??
<blank> == 0
```

- bool 리스트를 인덱스처럼 전달

```python
# True인 인덱스만 추출
x[<blank>]   
```

```python
# x중 80보다 큰것만 필터링하려면?
x[ x>80 ]
```

##  다중조건 필터링
 - nd-array 내에서 파이썬 논리 연산자인 and, or 키워드는 사용 불가
 - 기호로 대체 해야함
     - & (and) 
     - | (or) 

```python
#짝수이면서 30미만인 것만 필터링

x[(x % 2 == 0) <blank> (x < 30)]
```

```python
# 30미만이거나 50 초과인 것만 필터링 해보세요

x[(x<30) <blank> (x>50)   ]
```
