## 2) groupby() - 데이터 그룹핑
- groupby(by=그룹핑 기준컬럼).통계함수() 형태로 그룹별 통계량 확인

```python
lemonade.groupby(by=['Location'])
```

```python
lemonade.groupby(by=['Location']).count()
```

```python
lemonade.groupby(by=['Location']).mean(numeric_only=True)
```

- location별 그룹핑 하여 특정열(Revenue)만의 통계량을 보고싶다면?
    - groupby(by=그룹핑 기준컬럼)['열이름'].통계함수()

```python
lemonade.groupby(by=['Location'])['Revenue'].max()
```

```python
lemonade.groupby(by=['Location'])['Revenue'].min()
```

## 4) 중복 데이터 제거하기
### duplicated() - 중복 여부 확인 
- 행단위로 체크

```python
lemonade.duplicated()
```

```python
lemonade.head(10)  #7월6일 데이터가 2건
```

```python
lemonade.duplicated().sum()
```

- Parameters
    - keep
        - ‘first’ : Drop duplicates except for the first occurrence. (default)
        - ‘last’ : Drop duplicates except for the last occurrence.
        -  False : Drop all duplicates.
     -  ignore_index
        - True : 원래 index유지안하고 새로 indexing
        - False: 원래 index유지 (default)

```python
#중복값 제거  
lemonade.drop_duplicates(ignore_index=True) #원래 index유지안하고 새로 indexing
```

```python
#원본에는 반영되지 않음
lemonade
```

```python
lemonade.drop_duplicates(ignore_index=False)   #원래 index유지
```

```python
#새로 indexing
#원본에 반영: inplace=??

lemonade.drop_duplicates(keep='first', inplace=True, ignore_index=True)   
```

```python
lemonade
```

```python
lemonade.duplicated().sum()
```
