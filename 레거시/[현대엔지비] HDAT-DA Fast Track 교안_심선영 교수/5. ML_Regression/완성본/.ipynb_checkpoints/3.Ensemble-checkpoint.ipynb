# 앙상블 모델 (Ensemble Model) I
- 여러개의 모델을 결합하여 사용하는 모델
- 머신러닝의 앙상블은 크게 대표적으로 **Bagging**과 Boosting 두 가지로 구분됨

## 1. 배깅 모델 (Bagging Model)
- 상대적으로 약한 Decision Tree를 조합해서 사용하는 기법 
- Bagging - Bootstrap Aggregation의 약자
- 샘플을 여러 번 뽑는 Bootstrap 과정을 거쳐 각 모델을 학습시킨 후 그 결과물을 집계(Aggregation)하는 방법
- 배깅 모델에는 **랜덤 포레스트**와 **엑스트라 트리 리그레션**이 있음




- Regression, Classification 문제를 모두 지원
- 회귀의 경우 각 단일 모델들의 결과를 평균내어 예측
- 분류의 경우 각 단일 모델들의 결과에 투표하여 예측

### 1) 랜덤 포레스트 (Random Forest)
- 결정트리(decision tree)를 n개 결합한 모델
- 데이터 분할 최적화를 위해 **부트스트랩** 기법을 이용함
#### 부트스트랩(bootstrap)
- n개의 결정트리가 서로 다른 데이터로 학습토록 하기 위하여 원본 데이터셋에서 중복을 허용하여 무작위로 m개의 데이터를 선택하는 기법




- https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html
    - n=100(초기값)

### 2) 엑스트라 트리 리그레션 (ExtraTreesRegressor)
- 랜덤 포레스트가 변형된 형태
- **Extremely Randomized** 학습법 (샘플수, feature 설정이 모두 random)
- 부트스트랩은 사용하지 않도록 (bootstrap=False)로 파라미터 초기값 설정
- 그 외는 랜덤 포레스트와 동일한 앙상블 모델
- 단 랜덤 포레스트는 데이터를 최적 분할하지만 엑스트라 트리는 임의 분할하여 랜덤 포레스트보다 속도가 빠름

- https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.ExtraTreesRegressor.html
    - n=100(초기값)

# 앙상블 모델 (Ensemble Model) II
- 여러개의 단일 모델을 결합하여 사용하는 모델
- 머신러닝의 앙상블은 크게 Bagging과 **Boosting** 두 가지로 구분됨

## 부스팅 모델 (Boosting Model)
- 상대적으로 약한 Decision Tree를 조합해서 사용하는 기법 
- 약한 예측 모형들의 학습 에러에 가중치를 두고, 순차적으로 다음 학습 모델에 반영하여 강한 예측모형을 만드는 것
- 부스팅 모델에는 대표적으로 **XGBoost**와 **LightGBM**이 있음



- Regression, Classification 문제를 모두 지원

### 1) XGBoost (eXtreme Gradient Boosting)
- Boosting 기법을 이용하여 구현한 대표적 알고리즘
- 성능과 자원 효율이 좋아서 많이 사용되는 알고리즘
- https://xgboost.readthedocs.io/en/stable/python/python_api.html#xgboost.XGBRegressor
    - n_estimators=100(초기값)
    

### 2) LightGBM (Light Gradient Boosting Model)
- Boosting 기법을 이용하여 구현한 최신 알고리즘
- 학습시간이 짧아 XGBoost 보다 속도가 빠름(light)
- 파라미터의 수가 매우 많음
- https://lightgbm.readthedocs.io/en/latest/Parameters.html
- 주요 parameters
    - num_iterations=100(초기값)     
    - max_depth = 1(초기값)
