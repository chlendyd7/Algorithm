# HR 예측 AI모델 개발

## 개요
- 탑승자의 심박수(Heart Rate: HR)를 예측하는 AI 모델 개발

## 분석 배경 
- 심박수는 차량 주행 과정에서 탑승자의 상태를 파악할 수 있는 대표적인 지표 중 하나
- 특정 주행 상태에서 탑승자의 심박수가 과도하게 높다면 이는 탑승자에게 불안한 주행일 가능성을 의미
- 반대로 특정 주행 상태에서 탑승자의 심박수가 너무 낮다면 이를 통해 탑승자가 졸음운전 등의 이유로 운전이 불가능한 상태임을 유추할 수 있음
- 따라서 차량 주행 상태와 심박수의 관계를 명확하게 밝혀낼 수 있다면, 탑승자에게 적정 심박수를 유지하게 하는 주행 알고리즘을 개발하거나, 탑승자의 상태를 모니터링하는 시스템을 개발할 수 있음

## 평가 지표
- MAE : Mean Absolute Error


## 데이터 셋 정보

### [Data Files]
- train.csv: 학습용 데이터 (Features, Target) - 100,000건 
- test.csv:  예측용 데이터 (Features) - 49,980건

### [Feature]
- ID : 샘플 별 고유 ID
- CT : 조향장치 칼럼 토크
- MS : 조향장치 모터 회전속도
- MC : 조향장치 모터 전류
- B_Depth : 브레이크압
- LAT_ACCEL : 횡가속도
- LONG_ACCEL : 종가속도
- E_Status : 엔진상태
    - 0 : Engine Stop, 1 : Part Load, 2 : Pull , 3: Fuel Cut off, 4 : Start, 5 : Idle Speed
- A_Depth : 엑셀 페달 깊이
- F_Economy : 하이브리드 차량 연비
- Inhibit : 기어봉 위치
    - P : Parking, R : Reverse, N : Neutral, D : Drive
- Elevation : 고도

### [Target]
- HR : 심박수

### (3) 아웃라이어 제거
- train셋에만 해당
- target  아웃라이어 제거

```python
# boxplotting을 통해 outlier 존재여부 확인
sns.boxplot(y="HR",data=train)   
plt.show()
```

- 아웃라이어 제거 방법
    - 1) 사분위수 범위(IQR = Q3-Q1)을 계산
    - 2) **Q1-1.5IQR**과 **Q3+1.5IQR**을 각각 max와 min으로 설정
    - 3) [min, max]를 벗어나는 데이터는 아웃라이어로 간주하여 제거

```python
# 1) 사분위수 범위(IQR = Q3-Q1)을 계산한다.
Q1 = train['HR'].quantile(0.25)
Q3 = train['HR'].quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
#2) Q1-1.5IQR과 Q3+1.5IQR을 각각 max와 min으로 설정한다.
MIN = Q1 - 1.5 * IQR
MAX = Q3 + 1.5 * IQR
print(MIN, MAX)
```

```python
# 3) [min, max]를 벗어나는 데이터는 이상치로 간주하여 제거 
#실제로는 train을 사용(아웃라이어 제거 안함)
train2=train[(train['HR']>MIN) & (train['HR']<MAX)]
train2
```

- 총 5,061개의 이상치가 제거되어 train data의 갯수가 94,939개가 됨

### (6) 스케일링
- 스케일링을 통해 변수들의 데이터 분포나 범위를 조정

**[주의] 스케일링 시 주의사항**
- fit (스케일링을 위한 계산 수행): train data
- transform(적용): train data & test data 

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

#object 선언
minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()
```

```python
#scaling된 xtrain을 별도 생성
Xtrain_ohe_scaled = Xtrain_ohe.copy()
```

```python
# MinMaxScaling
Xtrain_ohe_scaled[['B_Depth', 'A_Depth', 'F_Economy']] = minmax_scaler.fit_transform(Xtrain_ohe[['B_Depth', 'A_Depth', 'F_Economy']])

# StandardScaling
Xtrain_ohe_scaled[['CT', 'MS', 'MC', 'LAT_ACCEL','LONG_ACCEL', 'Elevation']] = standard_scaler.fit_transform(Xtrain_ohe[['CT', 'MS', 'MC', 'LAT_ACCEL','LONG_ACCEL', 'Elevation']])
```

```python
# test 셋에는 transform만 적용
Xtest_ohe_scaled = Xtest_ohe.copy()
Xtest_ohe_scaled[['B_Depth', 'A_Depth', 'F_Economy']] = minmax_scaler.transform(Xtest_ohe[['B_Depth', 'A_Depth', 'F_Economy']])
Xtest_ohe_scaled[['CT', 'MS', 'MC', 'LAT_ACCEL','LONG_ACCEL', 'Elevation']] = standard_scaler.transform(Xtest_ohe[['CT', 'MS', 'MC', 'LAT_ACCEL','LONG_ACCEL', 'Elevation']])
```

```python
Xtrain_ohe_scaled
```

```python
Xtest_ohe_scaled
```

## [step4] 모델 설계 
- 모델의 선택 및 학습, 예측, 평가 분석 모델 설계
 - 다양한 모델의 성능비교 실시 
    - 선형 모델 - LinearRegression 
    - 트리기반 회귀모델 - DecisionTreeRegressor
    - 트리기반 앙상블 모델 (배깅 기법) - RandomForestRegressor, ExtraTreesRegressor
    - 트리기반 앙상블 모델 (부스팅 기법) - XGBRegressor, LGBMRegressor

- 성능을 평가하기 위해 학습 데이터셋에서 20%를 활용해 검증 데이터셋을 생성
 - test_size = 0.2
 - random_state = 0

### (1) 훈련 및 검증 데이터 분할

```python
# 데이터 분할: test_size=0.2, random_state=0

X_train, X_val, y_train, y_val = train_test_split(Xtrain_ohe_scaled, ytrain, test_size=0.2, random_state=0)
```

### (2) 평가지표 확인
- 평가지표가 MAE이므로 skleran의 mean_absolute_error() 함수 사용

### (3) 알고리즘별 학습, 예측, 평가

#### 1) Linear Regression

```python
# Estimator 설정
lr_reg = LinearRegression()

#학습
lr_reg.fit(X_train, y_train)

#예측
lr_pred = lr_reg.predict(X_val)

#평가
print('Linear Regression 모델의 MAE : ', round(mean_absolute_error(y_val, lr_pred),3))
```

#### 2) Decision Tree Regressor

```python
# Estimator 설정
dt_reg = DecisionTreeRegressor()

#학습
dt_reg.fit(X_train, y_train)

#예측
dt_pred = dt_reg.predict(X_val)

#평가 
print('Decision Tree Regression 모델의 MAE : ', round(mean_absolute_error(y_val, dt_pred),3))
```

#### 3) Random Forest Regressor

```python
rf_reg = RandomForestRegressor(n_jobs=-1)
rf_reg.fit(X_train, y_train)
rf_pred = rf_reg.predict(X_val)
print('Random Forest 모델의 MAE : ', round(mean_absolute_error(y_val, rf_pred),3))
```

#### 4) Extra Trees Regressor

```python
et_reg = ExtraTreesRegressor(n_jobs=-1)
et_reg.fit(X_train, y_train)
et_pred = et_reg.predict(X_val)
print('Extra Trees 모델의 MAE : ', round(mean_absolute_error(y_val, et_pred),3))
```

#### 5) XGB Regressor

```python
xgb_reg = XGBRegressor(n_jobs=-1)
xgb_reg.fit(X_train, y_train)
xgb_pred = xgb_reg.predict(X_val)
print('XGBoost 모델의 MAE : ', round(mean_absolute_error(y_val, xgb_pred),3))
```

#### 6) LGBM Regressor

```python
lgb_reg = LGBMRegressor(verbose=-1, n_jobs=-1)   #학습 및 예측시의 상세한 과정에 대한 메세지 및 warning 출력 생략)
lgb_reg.fit(X_train, y_train)
lgb_pred = lgb_reg.predict(X_val)
print('LGBM 모델의 MAE : ', round(mean_absolute_error(y_val, lgb_pred),3))
```

**[정리] 가장 성능이 좋은 모델을 선정** 

```python
# 상기의 내용을 for문을 이용하여 고쳐보세요
# 1) Linear Regression ~ 6) LGBM Regressor

models = {
    'Linear Regression': LinearRegression(),
    'Decision Tree Regression': DecisionTreeRegressor(),
    'Random Forest': RandomForestRegressor(n_jobs=-1),
    'Extra Trees Regressor': ExtraTreesRegressor(n_jobs=-1),
    'XGB Regressor' : XGBRegressor(n_jobs=-1),
    'LGBM Regressor' : LGBMRegressor(verbose=-1, n_jobs=-1)   #학습 및 예측시의 상세한 과정에 대한 메세지 및 warning 출력 생략)
}

# 초기화: selected_estimator, pre_rmsle
selected_estimator = None
min_mae = float('inf')

for name, estimator in models.items():        
    estimator.fit(X_train, y_train)
    pred = estimator.predict(X_val)
    mae_ = mean_absolute_error(y_val, pred)
    print(f'[{name} 모델의 mae]: {round(mae_, 3)}')
    #print('{:} 모델의 MAE : {:.3f}'.format(name, mae_))

   # 최고 성능 모델 기록
    if mae_ < min_mae:
        selected_estimator = estimator
        min_mae = mae_
        print(f'selected estimator: {name}')
    print('\n')    
```

### (4) 하이퍼 파라미터 튜닝
- 선정된 모델의 하이퍼파라미터 튜닝
- GridSearchCV(Gridsearch Cross Validation) 활용
    - estimator, parameters, cv, scoring

##### [주의] 회귀모델에서의 scoring
- GridSearchCV는 score값이 클수록 우수한 파라미터라고 판단함
- 하지만 회귀모델의 Error지표는 score값이 작을수록 우수한 파라미터임
- 따라서 회귀모델에서는 scoring함수를 negative함수로 활용함
    - https://scikit-learn.org/stable/modules/model_evaluation.html

```python
# 1) 파라미터 후보군 설정
# 많은 하이퍼파라미터를 대상으로 진행할 수 있지만, 시간이 오래 걸리므로 단계적으로 진행
parameters ={'min_samples_leaf' : [1,2],   #defalut =1
             'random_state':[37,156]}      #defalut = None
```

```python
# 2) GridSearchCV 설정: 
from sklearn.model_selection import GridSearchCV
grid_reg = GridSearchCV(selected_estimator, parameters,cv=5,
                           scoring='neg_mean_absolute_error') 
```

```python
# 3) GridSearchCV estimator로 학습: 
grid_reg.fit(Xtrain_ohe_scaled, ytrain)
```

```python
# 4) GridSearchCV 결과 확인
print('최적 하이퍼 파라미터 : ', grid_reg.best_params_)
```

```python
# 5) GridSearchCV score 확인
print('최적 하이퍼 파라미터 score: ', -1*grid_reg.best_score_)
```

- 'min_samples_leaf': 1, 'random_state': 156 에 고정
- 다른 파라미터를 추가 튜닝할 수 있으나 시간 관계상 생략
- 현재까지의 튜닝 결과만 반영 but...

- [주의] 하이퍼 파라미터 튜닝은 아래와 같이 다양한 파라미터에 대해 충분히 수행해 주어야 성능향상의 효과를 볼 수 있음
