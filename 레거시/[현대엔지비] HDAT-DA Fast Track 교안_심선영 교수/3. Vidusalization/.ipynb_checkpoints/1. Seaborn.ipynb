## 실습할 그래프 종류
* **1. 빈도 분포 그래프 (1) - sns.countplot(한변수-질적)**
* **2. 빈도 분포 그래프 (2) - sns.histplot(한변수-양적)**
* **3. 아웃라이어 체크 - sns.boxplot(한변수-양적)**
* **4. 두 변수의 상관관계 - sns.scatterplot(두변수-양적)**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings as wn
wn.filterwarnings('ignore')
```

```python
tips=pd.read_csv("tips.csv")
tips
```

## 1. 빈도 분포 그래프 (1) - countplot()
- sns.countplot(**x=** , data=None, ..)
  * **이산형 변수(질적 변수)** 에 대해 **빈도분포** 출력

```python
# 요일별(x) 건수를 bar롤 출력

sns.countplot(x = "day", data = tips)
plt.show()
```

* 흡연 여부로 구분해 보려면? - **구분자 hue**

```python
sns.countplot(x = "day", data = tips, <blank> = "smoker")
plt.show()
```

```python
# 변수 tip에 대해서 countplot을 수행해 보자!
```

## 2. 빈도 분포 그래프 (2) - histplot()
- sns.histplot(x=, data= )
   * **연속형 변수(양적 변수)** 에 대해 **빈도분포** 출력

```python
# tip변수에 대한 histplot

sns.<blank>(x = "tip", data = tips)
plt.show()
```

## 3. 박스플롯
- sns.boxplot(x=, data= )  /  sns.boxplot(y=, data= )  
- **연속형(양적) 변수** 의 **사분위수 분포** 를 boxplotting ( **극단치** 도 함께 표현)

[관련용어]
- 백분위 수 : 데이터를 백등분 한 것
- 사분위 수 : 데이터를 사등분 한 것
  - 제 1사분위 수 (Q1) : 전체 데이터 중 하위 25%에 해당하는 값
  - 제 2사분위 수 (Q2) : 전체 데이터 중 하위(상위) 50%에 해당하는 값 = 중간값
(median)
  - 제 3사분위 수 (Q3) : 전체 데이터 중 하위 75% (상위 25%)에 해당하는 값
  - 제 4사분위 수 (Q4) : 전체 데이터 중 하위 100% (가장 상위)에 해당하는 값
- IQR (사분위 범위) :  Q3 - Q1 (데이터의 중간 50%)


[그리는 방법]
- 1.전체 데이터에서 각 사분위수를 계산한다.(Q1,Q2,Q3,Q4)
- 2.그래프에서 제1 사분위와 제3 사분위를 밑변으로 하는 직사각형을 그리고, 제 2사분위에 해당하는 위치(중앙값)에 선분을 긋는다.
- 3.사분위수 범위(IQR = Q3-Q1)을 계산한다.
- 4.Q3+1.5IQR보다 크거나, Q1-1.5IQR의 위치에 max와 min을 표시하고 그보다 바깥쪽의 데이터는 **극단치**로 간주한다.

```python
sns.boxplot(x = "tip", data = tips)
plt.show()

# x만 지정한 경우
```

```python
sns.boxplot(y = "tip", data = tips)
plt.show()
```

```python
tips
```

```python
# 요일별로 구분하여 보고싶다면?

sns.boxplot(y = "tip", data = tips, hue='day')
plt.show()
```

## 4. 산점도 그래프(scatter)
- sns.scatterplot(**x=, y=**, data=None, ..)
- **연속형(양적) 변수** 간 **상관관계** 확인

```python
# 산점도만 출력
sns.scatterplot(x = "total_bill", y = "tip", data = tips)
```

[산점도와 회귀선을 함께 출력]
- sns.regplot()
  + fit_reg = True : 회귀선이 나오지 않도록 설정 (default)    
  + fit_reg = False: 회귀선이 나오지 않도록 설정 

```python
sns.regplot (x = "total_bill", y = "tip", data = tips)
```
