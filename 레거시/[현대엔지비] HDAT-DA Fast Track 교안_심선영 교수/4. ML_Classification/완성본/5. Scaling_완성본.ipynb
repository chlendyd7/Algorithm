# **Feature - 양적변수** 스케일링
- 서로 다른 변수 값의 범위를 일정한 수준으로 맞추는 것
- 1) Standard Scaling (표준화)
- 2) MinMax  Scaling (정규화, Normalization)

## 1) Standard Scaling (표준화)
- 데이터를 평균이 0이고 분산이 1인 값으로 변환하여 표준정규분포가 되도록 함
*   StandardScaler()

```python
import numpy as np
import pandas as pd
```

- 붓꽃 데이터 셋을 로딩

```python
iris = pd.read_csv('iris.csv')
iris
```

```python
#feature만 추출
iris_df=iris.drop('target', axis=1)
iris_df
```

- feature들의 평균 값

```python
iris_df.mean()
```

- feature들의 분산 값

```python
iris_df.var()
```

- StandardScaler class import

```python
from sklearn.preprocessing import StandardScaler
```

- StandardScaler객체 생성

```python
scaler = StandardScaler()
```

- StandardScaler객체의 fit( ) 과 transform( ) 호출

```python
scaler.fit(iris_df)
```

```python
iris_scaled = scaler.transform(iris_df)
iris_scaled
```

- transform( )시 scale 변환된 데이터 셋이 numpy ndarry type으로 반환되어 이를 DataFrame으로 변환

```python
iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris_df.columns)
iris_df_scaled 
```

- feature들의 평균 값

```python
iris_df_scaled.mean()
```

- feature들의 분산 값

```python
iris_df_scaled.var()
```

## 2) MinMax  Scaling (정규화)
- 데이터를 0과 1사이의 값으로 변환함 
* MinMaxScaler()

```python
from sklearn.preprocessing import MinMaxScaler
```

* **MinMaxScaler객체 생성**

```python
scaler = MinMaxScaler()
```

* **MinMaxScaler 로 데이터 셋 변환. fit() 과 transform() 호출** 

```python
scaler.fit(iris_df)
```

```python
iris_scaled = scaler.transform(iris_df)
iris_scaled
```

- feature들의 최소 값

```python
iris_scaled.min(axis=0)
```

- feature들의 최대 값

```python
iris_scaled.max(axis=0)
```

## 3) Scaler 이용시 주의사항 
- fit(): 스케일링을 위한 기준정보 설정 (ex. 데이터 세트의 최대값/최소값 설정)
    - train 셋에만 적용
- transform() : 설정된 정보를 바탕으로 데이터 변환(scaling)
    - train 셋 / test셋 모두 적용

- 즉 train셋으로 fit()된 스케일링 기준 정보를 그대로 test셋에 적용!
- 만약 test셋으로 다시 fit()하게 되면 스케일링 기준정보가 서로 달라짐

### [예제]
- train 셋: 0,1,2,...,10
- test 셋: 0,2,..,5

```python
from sklearn.preprocessing import MinMaxScaler
```

* 주의) Scaler클래스의 fit(), transform()은 2차원 이상 데이터만 가능
    - reshape(-1, 1)로 차원 변경

```python
train_array = np.arange(0, 11)
train_array
```

```python
test_array =  np.arange(0, 6)
test_array
```

```python
# MinMaxScaler객체 생성
scaler = MinMaxScaler()
```

- train_array로 fit()하게 되면 train_array 데이터의 최소값이 0, 최대값이 10으로 설정

```python
scaler.fit(train_array)
```

- 왜 에러가 날까??
- fit()함수는 2d의 X 데이터를 원함!!

```python
# 어떻게 수정해야 2d로 변형될까?
train_array = np.arange(0, 11).reshape(-1,1)
train_array
```

```python
test_array =  np.arange(0, 6).reshape(-1,1)
test_array
```

- transform() - 1/10 scale로 데이터 변환
    - 10이 1로 변환

```python
train_scaled = scaler.fit(train_array)
train_scaled
```

```python
test_scaled = scaler.transform(test_array)
test_scaled
```

- 질문1) 만약 test_darray =[0,1,2,........20]이라면 어떻게 transform될까?
- 질문2) 만약 test_darray =[-5,-4,-3,..0,..,3,4,5]라면 어떻게 transform될까?

- 만약 test셋으로도 fit을 하게되면 어떻게 될까?

```python
# test 셋: fit/trainsform
scaler.fit(test_array)
test_scaled = scaler.transform(test_array) # 5->1로 변환 
```

```python
# test 셋 확인
print('원본 test_array 데이터:\n', test_array)
print('Scale된 test_array 데이터:\n', test_scaled)
```

- 학습데이터와 테스트데이터가 동일한 범위값으로 변환되는 심각한 문제 발생!
- 따라서, fit은 train 셋에만 적용해야 함!
