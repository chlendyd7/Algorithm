# 앙상블 모델 (Ensemble Model) II
- 여러개의 단일 모델을 결합하여 사용하는 모델
- 머신러닝의 앙상블은 크게 Bagging과 **Boosting** 두 가지로 구분됨

## 부스팅 모델 (Boosting Model)
- 상대적으로 약한 Decision Tree를 조합해서 사용하는 기법 
- 약한 예측 모형들의 학습 에러에 가중치를 두고, 순차적으로 다음 학습 모델에 반영하여 강한 예측모형을 만드는 것
- 부스팅 모델에는 대표적으로 **XGBoost**와 **LightGBM**이 있음



- Regression, Classification 문제를 모두 지원

### 1) XGBoost (eXtreme Gradient Boosting)
- Boosting 기법을 이용하여 구현한 대표적 알고리즘
- 성능과 자원 효율이 좋아서 많이 사용되는 알고리즘
- https://xgboost.readthedocs.io/en/stable/python/python_api.html#xgboost.XGBRegressor
    - n_estimators=100(초기값)
    

### 2) LightGBM (Light Gradient Boosting Model)
- Boosting 기법을 이용하여 구현한 최신 알고리즘
- 학습시간이 짧아 XGBoost 보다 속도가 빠름(light)
- 파라미터의 수가 매우 많음
- https://lightgbm.readthedocs.io/en/latest/Parameters.html
- 주요 parameters
    - num_iterations=100(초기값)     
    - max_depth = 1(초기값) 

## 1. 라이브러리 import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt\

import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.model_selection import train_test_split  # 학습셋과 검증셋의 분리
from sklearn.metrics import r2_score   # 학습결과 성능평가
```

## 2. 데이터 셋 로드

```python
# 보스턴 주택 가격 예측
bostonDF=pd.read_csv('boston.csv')
bostonDF
```

## 3. Feature, Target 설정

```python
X_features=bostonDF.drop(["PRICE"], axis=1) 
y_target=bostonDF['PRICE']
```

## 4. xgboost와 lightgbm 기반 학습/예측

### 4.1) 알고리즘 import

```python
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
```

### 4.2) 학습/예측 

```python
# 1. estimator생성
xgb_reg = XGBRegressor()
lgb_reg = LGBMRegressor()
```

```python
# 2. 학습셋과 검증셋의 분리
X_train , X_val , y_train , y_val = train_test_split(X_features , y_target, test_size=0.3, random_state=156)
```

```python
# 3.학습
xgb_reg.fit(X_train , y_train)
lgb_reg.fit(X_train , y_train)
```

```python
# 4. 예측
pred_xgb = xgb_reg.predict(X_val)
pred_lgb = lgb_reg.predict(X_val)
```

### 4.3) 결과 확인

#### 모델별 R-square 확인

```python
print("R-sqaure of xgboost: ", round(r2_score(y_val, pred_xgb),4))
```

```python
print("R-sqaure of lightgbm: ", round(r2_score(y_val, pred_lgb),4))
```

#### 모델별 feature importances 확인

```python
print(xgb_reg.feature_names_in_)
print(xgb_reg.feature_importances_)
print(lgb_reg.feature_importances_)
```

```python
#데이터 프레임으로 만들기 - DecisionTree
boostingDF=pd.DataFrame(index=xgb_reg.feature_names_in_,
             data=xgb_reg.feature_importances_,
             columns=["xgboost_features"])

#데이터 프레임으로 만들기 - DecisionTree
boostingDF["LGBM_features"]=lgb_reg.feature_importances_
boostingDF
```

```python
#print(lgb_reg.feature_names_in_)
print(lgb_reg.feature_importances_)
```

```python
#데이터 프레임으로 만들기 - RandomForest
pd.DataFrame(index=xgb_reg.feature_names_in_,
             data=lgb_reg.feature_importances_,
             columns=["feature_importances"]).sort_values(by=["feature_importances"],ascending=False)
```
