# 2. Pandas 의 데이터 타입
## 1) Series
- 1차원 배열 
- 배열의 인덱스(index)와 값(values)으로 구성
- 생성시에는 index=, **data=**로 설정
- 조회시에는 sr.index, **sr.values**로 사용

```python
pd.Series(data=[17000, 18000, 1000, 5000])         
```

- index를 설정하지 않았으므로 임의의 값인 0,1,2,3으로 나옴

```python
pd.Series(index=[17000, 18000, 1900, 2000],
          data=["피자", "치킨","콜라", "맥주"])
```

- 'data='은 생략가능

```python
sr = pd.Series([17000, 18000, 1000, 5000])  
```

```python
sr = pd.Series([17000, 18000, 1000, 5000],
               index=["피자", "치킨", "콜라", "맥주"])
```

## 2) Dataframe

- 행과 열을 가지는 2차원 자료구조
- 행방향 이름: index 
- 열방향 이름: column

- 생성시에는 index=, columns=, **data=** 로 설정
- 조회시에는 df.index, df.columns=, **df.value**로 사용

```python
index = ['one', 'two', 'three']
columns = ['A', 'B', 'C']
values = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  
```

```python
df = pd.DataFrame(data=values, index=index, columns=columns)
df
```

- 생성된 데이터프레임으로부터 인덱스(index), 열(columns), 값(values)을 각각 출력해 보면..

```python
df.index
```

```python
df.columns
```

```python
df.values
```

# 3. DataFrame 의 생성
- DataFrame은 리스트(List), 시리즈(Series), 딕셔너리(dict), Numpy의 ndarray, csv로부터 생성할 수 있음
- pd.DataFrame()


## 1) 딕셔너리로부터 데이터프레임 생성 - {'key' : [value]}
- key: columns
- value: value
- index: 기본값

```python
data = {
    '학번' : ['1000', '1001'],
    '이름' : [ 'Steve', 'James'],
    '점수': [90.72, 78.09]
    }
```

```python
pd.DataFrame(data)
```
