## 3) loc - 원하는 행, 원하는 열 조회
* **df.loc[[원하는 행index], [원하는 열이름]]**

```python
lemonade.loc[[1,3,5], ['Date','Location']]   
```

```python
lemonade.loc[[0], ['Date','Location']]   
```

- 행만지정 (열은 모두 추출): df.loc[[원하는 행index]]
- 열만지정 (행은 모두 추출): df.loc[ : , [원하는 열이름]]

```python
lemonade.loc[[0]]
```

```python
lemonade.loc[0]
```

```python
lemonade.loc[[1,3,5]]  
```

```python
lemonade.loc[:, ['Date','Location']]
```

```python
lemonade[['Date','Location']] 
```

- loc로 연속된 행 추출시
    - 0-m행, 특정열 추출
    - **df.loc[0:m, [원하는 열이름]]**         
        - 주의: selection연산자(:)가 있는 부분은 대괄호 [] 하지 않음

```python
lemonade.loc[1:3,["Date", "Location"]] 
```

```python
lemonade.loc[1:3] 
```

#### [정리1]
- loc는 DataFrame의 함수가 아니라 속성임
- df.loc[] (df.loc()가 아니였음)
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html

#### [정리2]
- 원하는 열은 반드시 열이름으로 추출해야 함
- 원하는 열에 맵핑되는 숫자(index)로 추출할 수 없음

```python
lemonade.loc[[1,3,5], ['Date','Location']]   
```

## 4) iloc - 원하는 행과 열을 index로 조회
* **df.iloc[[원하는 행index], [원하는 열index]]**
- 행만지정 (열은 모두 추출): df.iloc[[원하는 행index]]
- 열만지정 (행은 모두 추출): df.iloc[:,[원하는 열index]]

```python
lemonade.iloc[[0,1,2], [1,2]] 
```

```python
lemonade.iloc[[0,1,2]] 
```

```python
lemonade.iloc[:, [1,3,5]] 
```

- iloc으로 연속된 구간 추출시
     - 0-m행, 0-n열 추출
    - **df.iloc[0:m+1, 0:n+1]** 
        - 주의1: "끝나는 index+1" (loc와 비교)
        - 주의2: selection연산자(:)가 있는 부분은 대괄호 [] 하지 않음

```python
lemonade.iloc[0:3, 0:2]   
```

```python
# 그럼 아래 코드로 첫 2열(0,1)을 추출할 수 있을까?
lemonade.iloc[0:2]   #안됨 (행으로 간주)
```

```python
lemonade.iloc[:, 0:2]
```

## 5) 조건 인덱싱 (Boolean Indexing과 유사)

```python
lemonade['Location'] == 'Beach'
```

* **만약 조건을 만족하는 lemonade의 행을 모두 출력하게 하려면?**
* **lemonade[조건]**

```python
lemonade[lemonade['Location'] == 'Beach']
```

```python
lemonade.loc[lemonade['Revenue']>45]
```

```python
lemonade.loc[lemonade['Revenue']>45, ['Date','Revenue']]
```

```python
lemonade.loc[ (lemonade['Revenue']>45) & (lemonade['Revenue']<100), ['Date', 'Revenue']]
```

```python
lemonade.loc[ (lemonade['Revenue']<45) | (lemonade['Revenue']>100), ['Date', 'Revenue']]
```

#### [정리]
- nd-darray처럼 df.loc[ ] 내에서도 AND, OR대신에 &, | 사용
