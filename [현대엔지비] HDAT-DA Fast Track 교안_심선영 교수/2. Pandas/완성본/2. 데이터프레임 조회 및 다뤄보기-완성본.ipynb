# 2. 데이터 다뤄보기

## 1) 열(column) 추가하기

```python
lemonade['Sold'] = 0 
```

```python
lemonade.head(3)
```

```python
lemonade['Sold'] = lemonade['Lemon'] + lemonade['Orange']
```

```python
lemonade.head(3)
```

```python
lemonade['Revenue'] = lemonade['Price'] * lemonade['Sold']
```

```python
lemonade.head(3)
```

## 2) 열(column) 제거하기


* **drop([열이름], axis=1(열방향이란 뜻))**

```python
lemonade.drop(['Sold'], axis=1)  
```

```python
lemonade.drop(['Date','Location'], axis=1)
```

```python
lemonade
```

* **drop([행번호(=index값)], axis=0(행방향이란 뜻)))**

```python
lemonade.drop([0], axis=0)
```

```python
lemonade
```

```python
# 2개이상의 행 삭제
lemonade.drop([0,1], axis=0)
```

### lemonade 자체에서 행과 열이 drop된 것은 아님. 그럴려면..
* **lemonade = lemonade.drop()**  혹은
* **inplace=True** 옵션을 써야함!

```python
lemonade.drop(['Orange'], axis=1, inplace=True)
```

```python
lemonade
```
