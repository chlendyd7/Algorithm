## 3) loc - 원하는 행, 원하는 열 조회
* **df.loc[[원하는 행index], [원하는 열이름]]**

```python
lemonade.loc[<blank>, ['Date','Location']]  
```

```python
lemonade.loc[[0], ['Date',<blank>]]   
```

```python
#index 0행 하나만 (모든열에 대해) 추출
lemonade.loc[[0]]  
```

```python
#이렇게 하면 어떻게 될까?
lemonade.loc[0]  #단일괄호 - Series로 출력
```

- 행만지정 (열은 모두 추출): df.loc[[원하는 행index]]
- 열만지정 (행은 모두 추출): df.loc[ : , [원하는 열이름]]

```python
#index 1,3,5행을 (모든열에 대해) 추출하고 싶다면
lemonade.loc[[1,3,5]]  
```

```python
#index 1,3,5행을 (모든열에 대해) 추출하고 싶다면
lemonade.loc[[1,3,5], <blank>]  
```

```python
#column 'Date','Location'을 (모든행에 대해) 추출하고 싶다면?
lemonade.loc[<blank>, ['Date','Location']] 
```

```python
#참고 - 또다른 방법
lemonade[['Date','Location']] 
```

- loc로 연속된 행 추출시
    - 0-m행, 특정열 추출
    - **df.loc[0:m, [원하는 열이름]]**         
        - 주의: selection연산자(:)가 있는 부분은 대괄호 [] 하지 않음

```python
#행-index 1~3, 열-["Date", "Location"]
lemonade.loc[<blank>,["Date", "Location"]] 
```

#### [정리1]
- loc는 DataFrame의 함수가 아니라 속성임
- df.loc[] (df.loc()가 아니였음)
- https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html

#### [정리2]
- 원하는 열은 반드시 열이름으로 추출해야 함
- 원하는 열에 맵핑되는 숫자(index)로 추출할 수 없음

```python
lemonade.loc[[1,3,5], ['Date','Location']]   
```

```python
# 위 코드를 이렇게 바꾸면? (column명도 숫자(index)로 변경)
lemonade.loc[[1,3,5],[0,1]]    #안됨
```

## 4) iloc - 원하는 행과 열을 index로 조회
* **df.iloc[[원하는 행index], [원하는 열index]]**
- 행만지정 (열은 모두 추출): df.iloc[[원하는 행index]]
- 열만지정 (행은 모두 추출): df.iloc[:,[원하는 열index]]

```python
# 행:0,1,2, 열:1,2를 원할 때
lemonade.iloc[[0,1,2], [1,2]] 
```

```python
# 행:0,1,2 만 추출 (열은 모두 추출)
lemonade.iloc[[0,1,2]] 
```

```python
# 열: 1,3,5만 추출 (행은 모두 추출) 
lemonade.iloc[<blank>, [1,3,5]] 
```

- iloc으로 연속된 구간 추출시
     - 0-m행, 0-n열 추출
    - **df.iloc[0:m+1, 0:n+1]** 
        - 주의1: "끝나는 index+1" (loc와 비교)
        - 주의2: selection연산자(:)가 있는 부분은 대괄호 [] 하지 않음

```python
lemonade.iloc[0:3, 0:2]   
```

## 5) 조건 인덱싱 (Boolean Indexing과 유사)

```python
#'Location'값이 'Beach'인 행 들만 filtering하는 조건
lemonade['Location'] == 'Beach'
```

* **상기 조건을 만족하는 lemonade의 행만 모두 출력하게 하려면?**
* **lemonade[조건]**

```python
lemonade[<blnak>]
```

```python
# 45<Revenue<100 조건일 때
lemonade[ (lemonade['Revenue']>45) <blank> (lemonade['Revenue']<100)]
```

```python
# 45<Revenue<100 조건일 때, ['Date', 'Revenue']열만 출력하려면?
lemonade.<blnak>[ (lemonade['Revenue']>45) & (lemonade['Revenue']<100), ['Date', 'Revenue']]
```

#### [정리]
- nd-darray처럼 df.loc[ ] 내에서도 AND, OR대신에 &, | 사용
