# 2. Pandas 의 데이터 타입
## 1) Series
- 1차원 배열 
- 배열의 인덱스(index)와 값(values)으로 구성
- 생성시에는 index=, **data=** 로 설정
- 조회시에는 sr.index, **sr.values** 로 사용

- index를 설정하지 않았으므로 임의의 값인 0,1,2,3으로 나옴

- 전달인자 이름을 생략하면'data', 'index' 순으로 매핑

```python
sr = pd.Series([17000, 18000, 1000, 5000],
               ["피자", "치킨", "콜라", "맥주"])
```

```python
sr
```

## 2) Dataframe

- 행과 열을 가지는 2차원 자료구조
- 행방향 이름: index 
- 열방향 이름: column

- 생성시에는 index=, columns=, **data=** 로 설정
- 조회시에는 df.index, df.columns=, **df.values**로 사용

```python
index_ = ['one', 'two', 'three']
columns_= ['A', 'B', 'C']
values_ = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  
```

# 3. DataFrame 의 생성
- DataFrame은 리스트(List), 시리즈(Series), 딕셔너리(dict), Numpy의 ndarray, csv로부터 생성할 수 있음
- pd.DataFrame()


## 1) 딕셔너리로부터 데이터프레임 생성 - {'key' : [value]}
- key: columns
- value: value
- index: 기본값

## 2) 외부 파일로 부터 데이터프레임 생성

- pd.read_csv()
