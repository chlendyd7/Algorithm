# Boolean Indexing

- nd-arry에 bool 리스트 (True, False)를 인덱스처럼 전달하여 True인 경우만 필터링

## 짝수인 경우만 필터링

```python
import numpy as np
```

- [1,100) 범위에서 10개의 정수 랜덤 추출

```python
x = np.random.randint(1, 100, size=10)
print(x)
```

- bool 리스트 생성 (True/False) :x==짝수?

```python
even_mask = (x%2==0)
print(even_mask)
```

- bool 리스트를 인덱스처럼 전달

```python
x[even_mask]    
```

```python
# True인 인덱스만 표시 - x[even_mask] 와 동일!
x[x%2==0]   
```

```python
# x중 30보다 큰것만 필터링하려면?
x[x>30]
```

##  다중조건으로 필터링
 - ndarray 내에서 파이썬 논리 연산지인 and, or, not 키워드는 사용 불가
 - 대신 기호로 대체함
     - & (AND) 
     - | (OR) 
     - ! (NOT) 

```python
x[(x % 2 == 0) & (x < 30)]
```

```python
# 30미만이거나 50 초과인 것만 필터링 해보세요

x[(x < 30)|(x > 50)]
```
