### (5) reshape 함수 활용
 - ndarray의 형태, 차원을 바꾸기 위해 사용
 - 원본의 형태는 바뀌지 않고 그대로 유지됨
 - 행방향 reshape: order='C' (기본값, 생략가능) 
 - 열방향 reshape: order='F' 

```python
x = np.arange(1, 16)
print(x)
```

```python
x.shape
```

```python
x.reshape(3, 5)
```

```python
x.reshape(5, 3)
```

```python
x.reshape(5,3,1)
```

```python
x.reshape(5,4)
```

```python
# 열방향으로 reshape하려면?
x.reshape(5,3, order='F')
```

```python
# x의 형태와 차원을 확인하자!
print(x)
print(x.ndim)
```

```python
# x의 차원을 실제 변환하려면 어떻게 해야할까?
x=x.reshape(5, 3)
```

```python
# x의 형태와 차원을 다시 확인
print(x)
print(x.ndim)
```
